```python
with open("initialdata.txt",encoding='utf-8') as f:
    data=f.read()
```

```python
chunks=data.split('\n\n')
```

```python
chunks[1]
```

```python
def parse_chunks(chunk):
    chunk=chunk.strip() #removes the whitespaces
    sep_chunk=chunk.split('\n') #separates all the attributes
    username=sep_chunk[0]
    post_line=sep_chunk[1].strip()
    followers_line=sep_chunk[2].strip()
    following_line=sep_chunk[3].strip()
    no_of_posts=post_line.split()[0]
    no_of_followers=followers_line.split()[0]
    no_of_following=following_line.split()[0]
    if len(sep_chunk>3):
        bio="\n".join(sep_chunk[4:])
    else:
        bio=""
    # print(username,no_of_posts,no_of_followers,no_of_following,bio)
        
    return {'username':username,'no_of_posts':no_of_posts,'no_of_followers':no_of_followers,'no_of_following':no_of_following,'bio':bio}
```

```python
all_chunks={}
for chunk in chunks:
    parsed_chunk=parse_chunks(chunk)
    all_chunks.append(parsed_chunk)
```
